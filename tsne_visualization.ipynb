# Visualización 3D con t-SNE y Plotly

Este notebook utiliza t-SNE para la reducción de dimensionalidad y visualizaciones 3D interactivas con Plotly.

```python
# Importar Librerías Necesarias
import pandas as pd
import numpy as np
from sklearn.manifold import TSNE
import plotly.express as px
```

```python
# Cargar y Preparar los Datos
df = pd.read_csv('ecommerce_data_1M.csv')
df = df.dropna()  # Eliminar valores nulos
features = ['Age', 'AnnualIncome', 'TimeOnSite', 'ItemsInCart', 'TotalSpending']
X = df[features]
```

```python
# Aplicar t-SNE para Reducción de Dimensionalidad
tsne = TSNE(n_components=3, random_state=42)
X_embedded = tsne.fit_transform(X)
df_embedded = pd.DataFrame(X_embedded, columns=['Dim1', 'Dim2', 'Dim3'])
```

```python
# Visualización 3D con Plotly
fig = px.scatter_3d(df_embedded, x='Dim1', y='Dim2', z='Dim3',
                  color=df['Gender'],
                  title='Visualización 3D con t-SNE')
fig.show()
```
